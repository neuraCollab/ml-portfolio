# Confussion matrix
![Confusion Matrix](./image1.png)

TP - модель правильно предсказала положительный класс
TN - модель правильно предсказада отрицательный класс
FP - ошибка I рода ("ложная тревога"): 
    "Пациент здоров, но модель сказала болен"
FN - ошибка II рода ("пропущенная угроза"):
    "Пациент болет, но модель сказала здоров"

Доля всех правильных предсказаний
Accurasy (точность) = TP + TN / TP + TN + FP + FN

Из всех кого мы названили больными -- сколько действительно больны?
Precision (точность положительного класса) = TP / TP + FP 

Из всех действительно больных сколько мы нашли?
Recall (полнота) = TP / TP + FN 

Из всех действительно здоровых -- сколько мы правильно определили?
Specificity = TN / TN + FP 

Гармоническое среднее Precission и Recall
F1_score = 2* (Presision * Recall / Presision + Recall)
